We have a **single** loop variable `i` that takes the values 0, 1, and 2. But the value of "i" wasn't directly needed...we only used the loop variable to define `current_title` and `current_rating`, which is what we actually used in the body of the loop.

In some sense, we had **two** loop variables ("current_title" and "current_rating"), although we explicitly defined only one.

The `zip` function actually allows us to explicitly define two loop variables. This is demonstrated below:

````python
good_movies = [] # create empty list

cut_off = 7.0

movie_titles = ["The Umbrellas of Cherbourg", "M3gan", "Psycho"]
movie_ratings = [7.8, 6.3, 8.5]

for current_title, current_rating in zip(movie_titles, movie_ratings):
    if current_rating >= cut_off:
        good_movies.append(current_title)

print(good_movies)
````

Perform this task within a "for" loop, using the `zip` function to create two loop variables that take values from the lists "movie_names" and "runtimes". Each list should contain the titles of the movies belonging to a given category.

1. Create an array of x-values (`vec_x`) with 100 evenly spaced points between 0 and 2π using `np.linspace()`.

2. Create three y-vectors:
   - `vec_y1` using `np.sin(vec_x)`
   - `vec_y2` using `np.cos(vec_x)`
   - `vec_y3` using `np.sin(vec_x)` + `np.cos(vec_x)`

3. Create a list called `list_vectors` containing all three vectors: [`vec_y1`, `vec_y2`, `vec_y3`]

4. Write a for-loop that iterates through `list_vectors` and plots each y-vector against `vec_x` on the same plot.

5. Add axis labels, a title, and a legend (see the command `plt.legend`)

**Note**: The final output should be a single 2D plot showing all three functions plotted against the same x-values. It should look something like this:

<img src="figures/trig_functions.png" alt="drawing" width="350"/>

````python
# Your code here
import numpy as np
import matplotlib.pyplot as plt 
vec_x = np.linspace(0, 2*np.pi, 100)

vec_y1 = np.sin(vec_x)
vec_y2 = np.cos(vec_x)
vec_y3 = np.sin(vec_x) + np.cos(vec_x)

list_vectors = [vec_y1, vec_y2, vec_y3]
legend = ["sinx(x)", "cos(x)", "sin(x) + cos(x)"]


for vec, leg in zip(list_vectors, legend):
    plt.plot(vec_x, vec, label=leg)
plt.xlabel("$x$")
plt.ylabel("$y$")
plt.title("Trig Functions")
plt.show()
    

````

Below, the setup code creates a random list with positive and negative numbers. Then 3 randomly chosen entries of the list are replaced with the strings "missing", "N/A", and "corrupted data". Our goal is to compute (i) the sum of all numbers in the list (ignoring the strings), and (ii) the product of all **positive** numbers in the list (again, ignoring the strings).

Write a "for" loop with **two** "if / continue" pairs that computes these two values. The behavior inside the loop should follow these steps:

1. Check if the current element of `random_list` has the type "str". If so, skip to the next pass through the loop.

2. Add the current element of `random_list` to `total_sum`, overwriting the old value of `total_sum`.

3. Check if the current element of `random_list` is less than or equal to zero. If so, skip to the next pass through the loop.

4. Multiply the current element of `random_list` to `positive_product`, overwriting the old value of `positive_product`.


````python
# Setup code
import numpy as np
N = 15
random_list = list(np.random.uniform(low = -5, high = 10, size = N))
i,j,k = np.random.choice(range(N), size = 3, replace = False)
random_list[i] = "missing"
random_list[j] = "N/A"
random_list[k] = "corrupted data"

total_sum = 0
positive_product = 1

# Your for loop here
for value in random_list:
    if type(value) == str:
        continue
    total_sum = total_sum + value
    if value <= 0:
        continue
    positive_product = positive_product*value


print(total_sum)
print(positive_product)
````
